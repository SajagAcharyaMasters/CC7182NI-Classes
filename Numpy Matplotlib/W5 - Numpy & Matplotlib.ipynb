```python
import numpy as np
import matplotlib.pyplot as plt
fig, ax = plt.subplots(3, 3)
r = 4

def circle(spacing, px, py):
    a = np.linspace(0, 2 * np.pi * r - 30, spacing)
    ax[px, py].plot(x:=r * np.cos(a), y:= r * np.sin(a))
    ax[px, py].set_ylim(-5, 5)
    ax[px, py].set_xlim(-5, 5)

m = np.arange(9)
m[0] = 200
m = m.reshape(3,3)
for i in range(3):
    for j in range(3):
        circle(m[i, j], i, j)

plt.show()

```

#### Problem: divisible by 3 & 4

```python
import numpy as np
import matplotlib.pyplot as plt
def matrix_plot(m):
    fig, ax = plt.subplots()
    ax.imshow(m, cmap='Pastel1_r')
    shape = m.shape
    for i in range(m.size):
        y, x = divmod(i, shape[0])
        ax.text(x, y, i, size=12, ha='center', va='center')
    return fig
m = np.arange(25, dtype=np.int8).reshape(5, 5)
mask = m % 3 == 0
mask &= m % 4 == 0
fig = matrix_plot(mask)
plt.show()
```

Basically the same as divisible by 3 * 4, which is 12

```python
fig = matrix_plot(np.where(m % 12 == 0, True, False))
plt.show()
```

#### Ellipse Mask
**Formula:** (x − x0)^2 / a2 + (y − y0)^2 / b2 = 1

```python
import matplotlib.pyplot as plt
img = plt.imread('parrots.PNG')
fig, ax = plt.subplots(layout='tight')
h,w = img.shape[:2]
y, x = np.indices((h, w))
# Same concept as
# x, y = np.arange(w), np.arange(h).reshape(h, 1)
x0, y0 = w/2, h/2
# Compute the ellipse equation and take the invalid values (The mask will cover outside the ellipse)
ellipse_mask = ((x - x0)**2 / x0**2 + (y - y0)**2 / y0**2) > 1
# Apply
img[ellipse_mask] = [0,0,0,1]
ax.imshow(img)
plt.show()
```
