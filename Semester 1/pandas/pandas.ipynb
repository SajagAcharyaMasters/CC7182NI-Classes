# DF: Wrangling col

## create

```python
import pandas as pd
col = 'ABCD'
idx = range(1, 4)
df = pd.DataFrame(
    data = { c: [c + str(r) for r in idx] for c in col },
    index = idx
)
print(df)
```

```python
df['list'] = [ 4, 5, 6 ]
df['range'] = range(3)
df['scalar'] = 'new'
df['split'] = list('new')
print(df)
```

## rename columns

```python
df.rename(
    columns = {
        'A' : 'Ant',
        'B' : 'Bat',
    },
    inplace = True, # Saves to the original instance, not a new one
)
print(df)
```

## reading tsv

```python
import datetime as dt
import pandas as pd
df = pd.read_csv(
    './floorsheet.tsv',
    encoding = 'utf-8',
    delimiter = '\t',
    index_col = ['ts'],
)
df['day'] = df.index.map(
    lambda r: dt.datetime.strptime(r[:10], '%Y-%m-%d')
)
print(df['day'])
```

# DF: Missing data

## series

```python
## create
import pandas as pd
import numpy as np
s = pd.Series([1, float('inf'), float('nan'), None, 'text'])
print(s)
print(s[s.isna()]) # masking
print(s.dropna())
```

## drop how

```python
import numpy as np
import pandas as pd

df = pd.DataFrame([
    [1, np.nan, 2, np.nan],
    [2, 3, 5, np.nan],
    [np.nan, 4, 6, np.nan]
])
print(df.dropna(axis='columns', how='all'))
print(df.dropna(axis='rows', thresh=3))
```

## handling with fill

```python
import numpy as np
import pandas as pd
s = pd.Series([1, np.nan, 2, None, 3], index=list('abcde'))
print('s.fillna(0)')
print(s.fillna(0))
print('s.ffill()')
print(s.ffill())
print('s.bfill()')
print(s.bfill())
```
