```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Linear fit using least squares method
def linear_fit_sq(x, y):
    xm = x.mean()
    ym = y.mean()

    numerator = np.sum([(x[i] - xm) * (y[i] - ym) for i in range(len(x))])
    denominator = np.sum([(x[i] - xm) ** 2 for i in range(len(x))])

    m = numerator / denominator
    b = ym - m * xm

    print(f'{m}x + {b:.3f}')
    line_y = m * x + b
    plt.plot(x, line_y)
    plt.scatter(x, y)
    plt.title('Linear fit - Least Squares')
    plt.tight_layout()
```

```python
x = np.array([2, 4, 6, 8, 10])
y = np.array([3.2, 3.8, 5.3, 5.7, 7.5])
linear_fit_sq(x, y)
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 6, 7, 8, 9])
linear_fit_sq(x, y)
```
