# Dataframes
there is columns and index like in spreadsheet

```python
import pandas as pd
color = pd.Series({
    'R' : 'red', 'Y' : 'yellow', 'G' : 'green',
})
df = pd.DataFrame({
    'color' : color,
    'signal' : {'R' : 'stop', 'Y' : 'wait', 'G' : 'go'},
    'person' : {'R' : 'go', 'Y' : 'stop', 'G' : 'stop'}
})
print(df)
```

## Structure

```python
print(df.index)
print(df.columns)
print(df.columns.values)
```

## Column

```python
print(df['signal'])
print(type(df.signal)) # lookout for name collision
print(df.signal.to_dict())
```

## Numpy ndarray

```python
print(df.values)
print(type(df.values))
```

## transpose

```python
print(df.T)
```

# DF: Wrangling col

## create

```python
import pandas as pd
col = 'ABCD'
idx = range(1, 4)
df = pd.DataFrame(
    data = { c: [c + str(r) for r in idx] for c in col },
    index = idx
)
print(df)
```

```python
df['list'] = [ 4, 5, 6 ]
df['range'] = range(3)
df['scalar'] = 'new'
df['split'] = list('new')
print(df)
```

## rename columns

```python
df.rename(
    columns = {
        'A' : 'Ant',
        'B' : 'Bat',
    },
    inplace = True, # Saves to the original instance, not a new one
)
print(df)
```

## reading tsv

```python
import datetime as dt
import pandas as pd
df = pd.read_csv(
    './floorsheet.tsv',
    encoding = 'utf-8',
    delimiter = '\t',
    index_col = ['ts'],
)
df['day'] = df.index.map(
    lambda r: dt.datetime.strptime(r[:10], '%Y-%m-%d')
)
print(df['day'])
```
