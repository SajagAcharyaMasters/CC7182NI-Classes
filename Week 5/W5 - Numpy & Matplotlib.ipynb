# Plotting

```python
import numpy as np
import matplotlib.pyplot as plt
x = [1, 3, 4, 5]
y = [3, 6, 9, 12]
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xlabel('X')
i=2
ax.set_ylabel(r'$\sum \frac{1}{x^%d}$'%i) #Fun rendering thing in matplotlib
ax.set_title('My plot')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.arange(101)
fig, ax = plt.subplots()
for i in range(3):
    ax.plot(x, x ** i, label=f'$x^{i}$')
ax.set_xlabel('X')
ax.set_ylabel('$X^Y$')
ax.set_ylim(0, 100)
ax.set_title('My plot')
ax.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 2)
x = np.arange(0, 7, .25)
ax[0].plot(x, y:=np.sin(x * np.pi), label=f'cos(x)')
print(x)
ax[0].legend()
ax[1].plot(x, y:=np.tanh(x * np.pi), label=f'+tanh(x)')
ax[1].plot(x, y:=-np.tanh(x * np.pi), label=f'-tanh(x)')
ax[1].legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 2)
x = np.linspace(0, 2 * np.pi, 100)
ax[0].plot(x, y:=np.cos(x**2), label=f'$cos(x^2)$')
ax[0].legend()
ax[1].plot(x, y:=np.tanh(x), label=f'$+tanh(x)$')
ax[1].plot(x, y:=-np.tanh(x), label=f'$-tanh(x)$')
ax[1].legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
fig, ax = plt.subplots(3, 3)
r = 4

def circle(spacing, px, py):
    a = np.linspace(0, 2 * np.pi * r - 30, spacing)
    ax[px, py].plot(x:=r * np.cos(a), y:= r * np.sin(a))
    ax[px, py].set_ylim(-5, 5)
    ax[px, py].set_xlim(-5, 5)

m = np.arange(9)
m[0] = 200
m = m.reshape(3,3)
for i in range(3):
    for j in range(3):
        circle(m[i, j], i, j)

plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt
m = np.arange(100).reshape(10, 10)
fig, ax = plt.subplots()
ax.imshow(m, cmap='Blues') # maps color according to matrix
for i in range(100):
    y, x = divmod(i, 10)
    ax.text(x, y, i, ha='center', va='center')
plt.show()
```

```python
def matrix_plot(m):
    fig, ax = plt.subplots()
    ax.imshow(m, cmap='Pastel1_r')
    shape = m.shape
    for i in range(m.size):
        y, x = divmod(i, shape[0])
        ax.text(x, y, i, size=12, ha='center', va='center')
    return fig
m = np.zeros(25, dtype=np.int8).reshape(5, 5)
m[1, 1] = 1
fig = matrix_plot(m)
plt.show()
```
