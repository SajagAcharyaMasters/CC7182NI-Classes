# bar

```python
import matplotlib.pyplot as plt
import pandas as pd

fig, ax = plt.subplots(layout='tight')
ax.bar(list('abcde'), [65, 61, 25, 22, 27])
plt.show()
```

## barh

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(layout='tight')
ax.barh(list('abcde'), [65, 61, 25, 22, 27])
plt.show()
```

## bar from data using pandas

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('../pandas/companies.tsv', delimiter='\t')
df['sector'].value_counts().plot(kind='barh', layout='tight')
plt.title('Company count by sector')
plt.show()
```

# pie

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(layout='tight')
ax.pie([99, 1], autopct='%1.1f%%', pctdistance=1.2)
ax.set_title('Time Spend On')
plt.legend(labels=['Slides', 'Meme'], bbox_to_anchor=(0.8, 1))
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('../pandas/companies.tsv', delimiter='\t')
count_data = df['sector'].value_counts()
ax = count_data.plot(kind='pie', layout='tight', labels=None, autopct='%1.1f%%', pctdistance=1.2)
plt.title('Company count by sector')
plt.legend(count_data.index, bbox_to_anchor=(2, 1))
plt.show()
```
